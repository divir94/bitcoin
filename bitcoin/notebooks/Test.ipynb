```python
import requests
from bs4 import BeautifulSoup
from collections import namedtuple
```

```python
CaseStatus = namedtuple('CaseStatus', ['receipt_num', 'title', 'description'])

def get_page(num):
    receipt_num = 'YSC{}'.format(num)
    url = 'https://egov.uscis.gov/casestatus/mycasestatus.do?appReceiptNum={}'.format(receipt_num)
    status = CaseStatus(receipt_num, title='', description='')
    
    try:
        response = requests.get(url=url, timeout=5)
        soup = BeautifulSoup(response.content, 'html.parser')
        content = soup.find('div', attrs={'class': 'appointment-sec'})
        title = content.find('h1').text
        description = content.find('p').text
        status = CaseStatus(receipt_num, title, description)
    except Exception as e:
        print 'Failed to scrape {}.\n{}'.format(receipt_num, e)
    return status


start = 1790200000
end = 1790310000
result = {}

for i in xrange(start, end):
    status = get_page(i)
    result[i] = status
    if i % 100 == 0:
        print 'Time: {}'.format(pd.datetime.now())
        print status
        print '=' * 30
```
